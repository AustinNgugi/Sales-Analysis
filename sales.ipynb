LETS FIND THE BREAK EVEN POINT ; BASICALLY WHERE DO THE DISCOUNTS START GENERATING LOSSES 

```python
whole_central=df[df['Region']=='Central']
```

```python
discount_analysis=whole_central.groupby('Discount')['Profit'].agg(
    count='count',
    total_profit='sum',
    avg_profit='mean',
    loss_making= lambda x :(x < 0).sum()
) 
```

```python
discount_analysis['pct_loss_making'] = (discount_analysis['loss_making'] / discount_analysis['count']) * 100
discount_analysis.sort_index()
```

Discovery: Central's discount policy has a clear breakeven point between 20% and 30%. Transactions discounted at 20% or below remain profitable on average (86%+ profitable). At 30% and above, the pattern reverses sharply — 92%+ of transactions become loss-making, and from 32% discount onward, essentially 100% of transactions lose money regardless of discount depth (32%–80% all show ~100% loss rates). This suggests Central's pricing does not absorb any discount beyond ~20–25% without falling below cost.

```python
region_discount=df.groupby(['Region','Discount'])['Profit'].agg(
    count='count',
    avg_profit='mean',
    loss_making= lambda x:(x < 0).sum() 
)
region_discount['pct_loss_making'] = (region_discount['loss_making'] / region_discount['count']) * 100
region_discount

```

DISCOVERY:The relationship between discount depth and loss is company-wide — every region shows a sharp profitability collapse once discounts exceed ~20-30%, with 90-100% of transactions becoming loss-making from that point on. However, Central is the only region offering an 80% discount tier, and it does so at high volume (300 transactions, more than any other region's deepest discount tier), all guaranteed loss-making. This makes Central disproportionately exposed to a discount-driven loss problem that technically exists everywhere, but which Central alone extended to its most extreme and costly level.

LOSS PREDICTION PREDICTOR 

```python
# What i'm doing rn is that i'm creating a loss column which is basically my target variable 
df['is_loss']=(df['Profit']<0).astype(int)
# These are basically the features required to make the prediction 
features=['Discount','Quantity','Category','Sub-Category','Region']
# X is our independent variable 
x=df[features].copy()
# y contains pur dependent variable 
y=df['is_loss']
# Lemme explain what exactly is going on here:So we have our categorical variables right ? Unfortunately our model doesn't understand categorical variable as a result  we need to convert it to a format that the machine can understand ie numrical values
# So we take the categorical variables and save them in a list
# We then go ahead and drop one categorical variable , lemme explain why.We have enough info to predict the other variable so we don't exactly need it 
# get_dummies is basically the translator 
X=pd.get_dummies(x,columns=['Category','Sub-Category','Region'],drop_first=True)




```

Lets check the class balance 

```python
y.value_counts()
```

```python
X.shape
```

```python
X.dtypes
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,y,test_size=0.2,random_state=43,stratify=y)
```

Lets perform GRID SEARCH

```python
# Lets import the model
from sklearn.tree import DecisionTreeClassifier
# Lets import the algorith used to search the best combination of hyper-parameters(We are using grid-search)
from sklearn.model_selection import GridSearchCV

param_grid = {
    'max_depth': [3, 4, 5, 6],
    'min_samples_split': [10, 20, 50],
    'min_samples_leaf': [5, 10, 20],
    'class_weight': [None, 'balanced']
}


tree_classifier=DecisionTreeClassifier(random_state=42)


# What this basically means is that take the model, take the parameters of interest , use 5 folds for cross validation,the metric to use for evaluation is f1
# Then use all the cores to perform the tasks in parallel( what n_jobs=1 means )
grid_search=GridSearchCV(
    tree_classifier,param_grid,cv=5,scoring='f1',n_jobs=1
)

grid_search.fit(X_train,Y_train)

 
```

Lets make the predictions and then evaluate the predictions 

```python
# Making the prediction

best_tree_classifier=grid_search.best_estimator_
y_pred=best_tree_classifier.predict(X_test)
```

```python
# Evaluating the prediction using the classification report and confusion matrix

from sklearn.metrics import confusion_matrix,classification_report,accuracy_score

cr=classification_report(Y_test,y_pred)
cm=confusion_matrix(Y_test,y_pred)
acs=accuracy_score(Y_test,y_pred)
```
